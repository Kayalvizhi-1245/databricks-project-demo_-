```python


from pyspark.sql.functions import *

df = spark.read.format("csv") \
.option("header",True) \
.schema(claims_schema) \
.load(f"{source_path}/claims")

df = df.withColumn(

    "ingestion_time",

    current_timestamp()

)

df.write \
.format("delta") \
.mode("overwrite") \
.save(f"{bronze_path}/claims")
```
