```python
from pyspark.sql.types import *

source_path="/Volumes/hospital_schema_management/hospital_schema/hospital_schemaevaluation/raw/"

bronze_path="/Volumes/hospital_schema_management/hospital_schema/hospital_schemaevaluation/bronze/"

validated_path="/Volumes/hospital_schema_management/hospital_schema/hospital_schemaevaluation/validated/"

quarantine_path="/Volumes/hospital_schema_management/hospital_schema/hospital_schemaevaluation/quarantine/"


patients_schema = StructType([

StructField("patient_id",IntegerType(),True),

StructField("patient_name",StringType(),True),

StructField("city",StringType(),True),

StructField("gender",StringType(),True),

StructField("age",IntegerType(),True),

StructField("admission_date",DateType(),True)

])


claims_schema = StructType([

StructField("claim_id",StringType(),True),

StructField("patient_id",IntegerType(),True),

StructField("claim_amount",DoubleType(),True),

StructField("claim_status",StringType(),True),

StructField("claim_date",DateType(),True),

StructField("hospital_name",StringType(),True)

])


def validate_schema(df, expected_schema):

    actual={

        f.name:str(f.dataType)

        for f in df.schema.fields

    }

    expected={

        f.name:str(f.dataType)

        for f in expected_schema.fields

    }

    missing=[]

    datatype=[]

    extra=[]

    for col in expected:

        if col not in actual:

            missing.append(col)

        elif actual[col] != expected[col]:

            datatype.append(

                f"{col} expected={expected[col]} actual={actual[col]}"

            )

    for col in actual:

        if col not in expected:

            extra.append(col)

    return missing,datatype,extra
```
